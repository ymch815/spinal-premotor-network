```python
# temp
import os
import re

# Specify the directory containing your files
directory = './frequency_modulation/'

def process_directory(directory):
    # Iterate over all files in the directory
    for filename in os.listdir(directory):
        # Full file path
        file_path = os.path.join(directory, filename)

        # Check if it's a file (not a directory)
        if os.path.isfile(file_path):
            # 1. Remove files like "*ccrate_X.txt"
            if re.match(r".*ccrate_X\.txt$", filename):
                print(f"Removing {filename}")
                os.remove(file_path)

            # 2. Rename files like "*ccrate_F.txt" to "*ccrate.txt"
            elif re.match(r".*ccrate_F\.txt$", filename):
                new_filename = re.sub(r"ccrate_F\.txt$", "ccrate.txt", filename)
                new_file_path = os.path.join(directory, new_filename)
                print(f"Renaming {filename} to {new_filename}")
                os.rename(file_path, new_file_path)

            elif re.match(r".*jointfreq_X\.txt$", filename):
                print(f"Removing {filename}")
                os.remove(file_path)

            # 2. Rename files like "*ccrate_F.txt" to "*ccrate.txt"
            elif re.match(r".*jointfreq_F\.txt$", filename):
                new_filename = re.sub(r"jointfreq_F\.txt$", "jointfreq.txt", filename)
                new_file_path = os.path.join(directory, new_filename)
                print(f"Renaming {filename} to {new_filename}")
                os.rename(file_path, new_file_path)

            elif re.match(r".*phasediff_X\.txt$", filename):
                print(f"Removing {filename}")
                os.remove(file_path)

            # 2. Rename files like "*ccrate_F.txt" to "*ccrate.txt"
            elif re.match(r".*phasediff_F\.txt$", filename):
                new_filename = re.sub(r"phasediff_F\.txt$", "phasediff.txt", filename)
                new_file_path = os.path.join(directory, new_filename)
                print(f"Renaming {filename} to {new_filename}")
                os.rename(file_path, new_file_path)

for dirpath, dirnames, filenames in os.walk(directory):
    print(f"Processing directory: {dirpath}")
    process_directory(dirpath)
```

```python
# clean parameters
import os
import pandas as pd
import glob

# Specify the directory where the CSV files are stored
directory = 'frequency_modulation/'

# Function to process each CSV file
def process_csv(file_path):
    # Read the CSV file
    df = pd.read_csv(file_path)

    # Check if the 'Variables' column exists and drop it
    if 'Variable' in df.columns:
        df = df.drop(columns=['Variable'])

    # Drop duplicates to keep only one row for each unique value of 'Repitition'
    df_unique = df.drop_duplicates(subset='Repetition')
    df_unique = df_unique.reset_index().iloc[:,2:]

    # Save the modified dataframe back to the CSV file
    df_unique.to_csv(file_path, index=False)
    print(f"Processed {file_path}")

def process_csvs(directory):
    # Find all CSV files that match the pattern "*parameters.csv"
    csv_files = glob.glob(os.path.join(directory, '*parameters.csv'))

    # Process each CSV file
    for csv_file in csv_files:
        process_csv(csv_file)
for dirpath, dirnames, filenames in os.walk(directory):
    print(f"Processing directory: {dirpath}")
    process_csvs(dirpath)
```
